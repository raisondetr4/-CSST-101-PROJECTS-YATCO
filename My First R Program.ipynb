**3. Numerical Operators**

```r
x <- 10
y <- 3
add <- x + y
sub <- x - y
mul <- x * y
div <- x / y
mod <- x %% y
exp <- x ^ y

print(add) #addition
print(sub) #subtraction
print(mul) #multiplication
print(div) #division
print(mod) #modulus
print(exp) #exponent


```

**7. Functions**

```r
addNumber <- function(a, b){
  return(a+b)
}
subNumber <- function(c, d){
  return(c-d)
}
mulNumber <- function(e, f){
  return(e*f)
}
divNumber <- function(g, h){
  return(g/h)
}

addNumber(3,7)
subNumber(10,18)
mulNumber(4,16)
divNumber(30,5)
```
