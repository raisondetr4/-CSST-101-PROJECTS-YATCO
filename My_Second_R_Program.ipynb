**14. ???**

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95,72,88,79,68),
  Grade = c(90,70,85,77,65)
)

students$Performance <- ifelse(
  students$Grade >= 80,
  "Good",
  ifelse(
    students$Grade >=75,
    "Passed",
    "Failed"
  )
)

students$Risk_Status <- ifelse(
  students$Attendance < 75 & students$Grade < 75,
  "High Risk",
  "Low Risk"
)

student_risk <- function(Attendance, Grade) {
  if (Attendance <75 & Grade < 75) {
    return("High Risk")
  } else if (Attendance < 85 | Grade < 80) {
    return("Moderate Risk")
  }else {
    return("Low Risk")
  }
}

students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

risk_counts <- table(students$Risk)
barplot(
  risk_counts,
  main ="Student Risk Classification",
  xlab = "Risk Level",
  ylab = "Number of students"
)

get_recommendation <- function (risk) {
  if (risk == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk == "Moderate Risk") {
    return("Academic Monitoring")
  } else {
    return("Continue regular monitoring")
  }
}

students$Recommendation <- mapply(get_recommendation, students$Risk)

View(students)
print(table(students$Risk))
```

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95,72,88,79,68),
  Grade = c(90,70,85,77,65)
)

students$Performance <- ifelse(
  students$Grade >= 80,
  "Good",
  ifelse(
    students$Grade >=75,
    "Passed",
    "Failed"
  )
)

students$Risk_Status <- ifelse(
  students$Attendance < 75 & students$Grade < 75,
  "High Risk",
  "Low Risk"
)

student_risk <- function(Attendance, Grade) {
  if (Attendance <75 & Grade < 75) {
    return("High Risk")
  } else if (Attendance < 85 | Grade < 80) {
    return("Moderate Risk")
  }else {
    return("Low Risk")
  }
}

students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

get_recommendation <- function (risk) {
  if (risk == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk == "Moderate Risk") {
    return("Academic Monitoring")
  } else {
    return("Continue regular monitoring")
  }
}

students$Recommendation <- mapply(get_recommendation, students$Risk)

student_id <- "S002"
student_info <- subset(students, Student_ID == student_id)

cat("Student_ID:", student_info$Student_ID, "\n")
cat("Attendance:", student_info$Attendance, "%\n")
cat("Grade:", student_info$Grade, "\n")
cat("Risk Level:", student_info$Risk, "\n")
cat("Recommendation:", student_info$Recommendation, "\n")

View(students)
print(table(students$Risk))
```
