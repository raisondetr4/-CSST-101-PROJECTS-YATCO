```r
Knowledge_Rules <- data.frame(
    Rule = c("R1", "R2", "R3", "R4", "R5", "R6", "R7", "R8"),
    IF_condition = c(
        "Fire reported",
        "Fire alert verified",
        "Confirmed incident",
        "Exit route blocked",
        "Confirmed incident AND Exit route blocked",
        "If your clothes catch on fire",
        "If there is smoke",
        "If fire is confirmed"
    ),
    THEN_conclusion = c(
        "Fire alert",
        "Confirmed incident",
        "Follow approved evacuation procedures",
        "Report blocked exit",
        "Recommend alternative approved exit",
        "Stop, drop and roll",
        "Stay low",
        "Call emergency services"
    ),
    stringsAsFactors = FALSE
)

Facts <- c(
    "Fire reported",
    "Fire alert verified",
    "Confirmed incident",
    "Exit route blocked",
    "Confirmed incident and exit route blocked",
    "If your clothes catch on fire",
    "If there is smoke",
    "If fire is confirmed"
)

Fired_Rules <- character(0)

Rule_Trace <- data.frame(
    Rule = character(),
    IF_condition = character(),
    THEN_conclusion = character(),
    Conclusion_Status = character(),
    stringsAsFactors = FALSE
)

repeat {
  New_Fact_Added <- FALSE

  for (i in seq_len(nrow(Knowledge_Rules))) {

    Rule_ID <- Knowledge_Rules$Rule[i]
    Condition <- Knowledge_Rules$IF_condition[i]
    Conclusion <- Knowledge_Rules$THEN_conclusion[i]

    if (Rule_ID %in% Fired_Rules){
      next
    }

    Required_Facts <- strsplit(
      Condition, "AND", fixed = TRUE
    ) [[1]]

    if (Condition_Met){
      Fired_Rules <- c(Fired_Rules, Rule_ID)

      if (!(Conclusion %in% Facts)) {
        Facts <- c(Facts, Conclusion)
        Status <- "New Conclusion Added"
        New_Fact_Added <- TRUE
      } else {
        Status <- "Duplicate Found"
      }
      Rule_Trace <- rbind(
        Rule_Trace,
        data.frame(
          Rule = Rule_ID,
          IF_condition = Condition,
          THEN_conclusion = Conclusion,
          Conclusion_Status = Status,
          stringAsFactors = FALSE
        )
      )
    }
  }
  if (!New_Fact_Added){
    break
  }
}

cat("KNOWLEDGE BASE:")
cat("\n")
cat("Forward Chaining:")
cat("\n")
print(unique(Facts))
cat("\n")
cat("Prevent Duplicate:")
cat("\n")
print(Rule_Trace, row.names = FALSE)
cat("\n")
cat("\n")
print(Fired_Rules)
```

**3-4. Reports**

```r
Conflicting_Pairs <- list(
  c("Fire reported", "No fire reported"),
  c("Fire alert verified", "Fire alert unverified"),
  c("Exit route blocked", "Exit route clear")
)

Conflicts <- sapply(Conflicting_Pairs, function(pair){
  all(pair %in% Facts)
})

cat("Conflicting Reports")
cat("\n")
if (any(Conflicts)) {
  print(Conflicting_Pairs[Conflicts])
} else {
  cat("No conflicting reports")
  cat("\n")
}

Missing_Information <- character(0)

if ("Fire reported" %in% Facts &&
   !any(c("Fire alert verified",
          "Fire alert unverified") %in% Facts)) {
    Missing_Information <- c(
      Missing_Information, "Fire report verification is missing"
    )
    }

if (!any(c("Exit route blocked",
           "Exit route clear") %in% Facts)) {
            Missing_Information <- c(
              Missing_Information, "Exit route condition is unknown"
  )
}

cat("\n")
cat("Missing Information:")
cat("\n")
if (length(Missing_Information) > 0) {
  print(Missing_Information)
} else {
  cat("No missing information detected.")
  cat("\n")
}

cat("\n")
cat("Incident Verification:")

if ("Confirmed incident" %in% Facts) {
  cat("Incident is confirmed.")
  cat("\n")
} else {
  cat("Incident is not confirmed.")
  cat("Report should not be treated as fact.")
}

Test_Scenarios <- list(
  list(
    Test_Number = "1. Unverified Fire Report",
    Facts = c("Fire reported"),
    Expected = "Unverified report, do not confirm incident"
  ),
  list(
    Test_Number = "2. Verified Fire",
    Facts = c("Fire reported", "Fire alert verified"),
    Expected = "Fire alert and confirmed incident"
  ),
  list(
    Test_Number = "3. Confirmed Fire with Blocked Exit",
    Facts = c("Confirmed incident", "Exit route blocked"),
    Expected = "Recommended alternative approved exits."
  ),
  list(
    Test_Number = "4. Conflicting Fire Reports",
    Facts = c("Fire reported", "No fire reported"),
    Expected = "Detect conflicting reports"
  ),
  list(
    Test_Number = "5. No Incident Reported",
    Facts = c("No fire reported"),
    Expected = "No confirmed incident, no fire concluded"
  )
)

Test_Results <- data.frame(
  Scenario = character(),
  OResults = character(),
  EResults = character(),
  stringsAsFactors = FALSE
)

for (Test in Test_Scenarios) {
  F <- Test$Facts
  Output <- character(0)

  if ("Fire reported" %in% F &&
    !("Fire alert verified" %in% F) &&
    !("Fire alert unverified" %in% F)){
      Output <- c(Output, "Report unverified")
    }

  if (all(c("Fire reported", "No fire reported") %in% F)){
    Output <- c(Output, "Conflicting reports detected")
  }
  if ("Confirmed incident" %in% F){
    Output <- c(Output, "Confirmed incident")
  } else {
    Output <- c(Output, "Incident not confirmed")
  }

  if (all(c("Confirmed incident",
            "Exit route blocked") %in% F)){
      Output <- c(
        Output, "Recommended alternative approved exit"
      )
    } else if (!any(c("Exit route blocked", "Exit route clear") %in% F)) {
      Output <- c(Output, "Route condition unknown")
    }

  if ("No fire reported" %in% F &&
    !("Fire reported" %in% F)){
      Output <- c(Output, "No fire reported")
    }

    Test_Results <- rbind(
      Test_Results, data.frame(
        Scenario = Test$Test_Number,
        OResults = paste(unique(Output), collapse = "; "),
        EResults = Test$Expected,
        stringAsFactors = FALSE
      )
    )
}

# Display results

cat("\n")
print(Test_Results, row.names = FALSE)
```
