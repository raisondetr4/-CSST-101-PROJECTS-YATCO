**Table 1. Debate Arguments**

```r
Debate_Arguments <- data.frame(
    ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
    Position = c("For", "For", "For", "Against", "Against", "Against"),
    Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity", "Dependency", "Assessment"),
    Claim = c("AI can provide students with immediate learning support.", "AI can help students complete routine academic tasks more efficiently.", "Allowing AI use can help students develop AI literacy.", "Undisclosed AI-generated work may create academic integrity concerns.", "Excessive AI use may reduce students independent problem-solving practice.", "AI use may make it more difficult to determine whether submitted work reflects student learning.")

)

Debate_Arguments
```

**Table 2. Evidence**

```r
Evidence <- data.frame(
  eID = c("E001", "E002", "E003", "E004", "E005", "E006"),
  ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  E_Type = c("Research", "Research", "Policy", "Policy", "Research", "Research"),
  Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong")
)

Evidence
```

**3. Connect Table Arguments and Evidence**

```r
Knowledge_Base <- merge(
  Debate_Arguments,
  Evidence,
  by = "ID"
)

print(Knowledge_Base)
```

**4. Convert Evidence**

```r
Get_Score <- function(Strength){

  if(Strength == "Strong"){
    return(3)
  } else if(Strength == "Moderate"){
    return(2)
  } else if(Strength == "Weak"){
    return(1)
  } else {
    return(0)
  }
}

Get_Score("Strong")
Get_Score("Moderate")
Get_Score("Weak")
Get_Score("Nye")
```

**5. Add Evidence Score**

```r
Knowledge_Base$Evidence_Score <- sapply(
  Knowledge_Base$Strength,
  Get_Score
)

print(Knowledge_Base)
```

**6. Evidence Score to Support**

```r
Classify_Support <- function(Evidence_Score){

  if(Evidence_Score == 3){
    return("Strongly Supported")
  } else if (Evidence_Score == 2){
    return("Moderately Supported")
  } else if (Evidence_Score == 1){
    return("Weakly Supported")
  } else {
    return("No Support")
  }
}

Classify_Support(1)
```

**7. Add Evidence Support**

```r
Knowledge_Base$Support_Level <- sapply(
  Knowledge_Base$Evidence_Score,
  Classify_Support
)

print(Knowledge_Base)
```

**8. Search by Topic**

```r
Search_Topic <- function(topic){

  Result <- subset(
    Knowledge_Base,
    Topic == topic
  )
  if(nrow(Result) == 0){
    cat("Topic not found")
  } else {
    print(Result)
  }

}
```
